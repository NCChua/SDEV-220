```python
# Exercise 4.1
secret = 3
guess = 8

if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
```

```python
# Exercise 4.2
small = True
green = False

if small and green:
    print("pea")
elif small and not green:
    print("cherry")
elif green and not small:
    print("watermelon")
else: 
    print("pumpkin")
```

```python
# Exercise 6.2
guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
    else:
        print("oops")
        break

    number += 1
```

```python
# Exercise 6.3
guess_me = 5

for number in range(10):
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
```
