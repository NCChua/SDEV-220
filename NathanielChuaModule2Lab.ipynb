```python
# Name: Nathaniel Chua
# file name: NathanielChuaModule2Lab.ipynb
#  Python app that will accept student names and GPAs and test if the student qualifies for either the Dean's List or the Honor Roll.
while True:
    last_name = input("Enter your last name or enter 'ZZZ' to quit. ")

    # Check for terminator
    if last_name == 'ZZZ':
        break
    
    # Gather all information from student
    first_name = input("Enter your first name: ")
    student_gpa = float(input("Enter your GPA: "))
    
    # Test for qualification of Dean's List or Honor Roll
    if student_gpa >= 3.5:
        print(f"Congratulations, {first_name}! You've made the Dean's List!")
    elif student_gpa >= 3.25:
        print(f"Congratulations, {first_name}! You've made the Honor Roll!")
    else:
        print(f"{first_name} {last_name} does not qualify.")
                        
```
